Bronze Customers

```python
%sql
CREATE OR REPLACE TABLE bronze_customers
USING DELTA
AS
SELECT *
FROM read_files(
  's3://capgemini-retail-etl-shiva/raw/customers_src_*.csv',
  format => 'csv',
  header => true
);
```
